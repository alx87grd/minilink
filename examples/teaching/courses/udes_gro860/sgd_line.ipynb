# Simple stochastic gradient descent algo

```
import numpy as np
import matplotlib.pyplot as plt
```

# True parameters

Here for this example, we are going to try to learn the two parameter of a line:

$y=ax+b$

```
a = 0.5
b = 3.0
```

# Data generation

Based on noisy data:

```
n = 1000
x = np.random.uniform(size=n) * 10 # samples x between 0 and 10
w = np.random.normal(size=n) * 0.5 # Normal noise

y = a * x + b + w
```

```
plt.scatter(x, y)
```

# Stochastic gradient descent

Exercise: Try to create a learning rate schedule leading to improved performance.

```
# Learning rate schedule
def nabla(i):
  return 0.02
```

```
w_hat = np.array([1.0, 1.0])

A = np.zeros(n)
B = np.zeros(n)

for i in range(n):
  phi = np.array([x[i],1.])
  y_hat = np.dot(w_hat, phi)
  error = y[i] - y_hat
  w_hat = w_hat + nabla(i) * error * phi

  # For ploting
  #print('a=',w_hat[0],'b=',w_hat[1])
  A[i] = w_hat[0]
  B[i] = w_hat[1]
```

```
plt.plot( np.arange(n), A , 'r')
plt.plot( np.arange(n), a * np.ones(n) , 'b')
```

```
plt.plot( np.arange(n), B , 'r')
plt.plot( np.arange(n), b * np.ones(n) , 'b')
```
