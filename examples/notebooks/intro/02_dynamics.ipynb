## Catalog plant

Pick a plant, set `params` / `x0`, simulate with the façade.


```python
import numpy as np
from minilink.dynamics.catalog.pendulum.cartpole import CartPole

plant = CartPole()
plant.params["m1"] = 2.0
plant.params["m2"] = 0.5
plant.params["lcg"] = 1.0
plant.x0 = np.array([0.0, 0.0, 0.2, 0.0])
plant.plot_diagram()
plant.compute_trajectory(tf=5.0)
plant.plot_trajectory()
```

## Abstraction: LTI and forced response

`LTISystem` is the linear textbook form $\dot x = A x + B u$. Catalog nonlinear
plants still expose the same `System` boundary.


```python
import numpy as np
from minilink.blocks.sources import Step
from minilink.dynamics.catalog.pendulum.cartpole import CartPole

plant = CartPole()
plant.params["m1"] = 2.0
plant.params["m2"] = 0.5
plant.params["lcg"] = 1.0
plant.x0 = np.array([0.0, 0.0, 0.0, 0.0])

step = Step(initial_value=0.0, final_value=1.0, step_time=1.0)
open_loop = step >> plant
open_loop.compute_trajectory(tf=6.0)
open_loop.plot_trajectory()
```
